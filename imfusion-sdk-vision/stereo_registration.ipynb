```python
import imfusion as imf
from imfusion import vision

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image set with calibration
stereo_images = imf.load("data/camera_registration.imf")
# Load marker borad configuration file
marker_config = imf.vision.marker_configuration.load_marker("data/camera_registration_marker_configuration.xml")
# calibration algorithm settings
settings = imf.vision.CameraCalibrationSettings()
```

```python
result = vision.stereo.register_camera(stereo_images[0], marker_config, settings)

print(f"\nRegistration results:")
print(f"MRE: {result}")
print(f"Left to Right Registration Matrix: {stereo_images[0].components.stereo_calibration.left_to_right_registration}")
```
