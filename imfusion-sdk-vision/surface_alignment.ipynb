## Surface Alignment Tutorial

This tutorial demonstrates how to use ImFusion's surface alignment functionality for aligning point clouds and meshes

```python
import imfusion as imf
from imfusion import vision # Surface alignment is in the vision module
```

```python
# Load two point clouds
pcs = imf.load("data/surface_alignment.imf")
```

```python
# Run registration
vision.align(pcs, use_gpu=False, icp_algorithm=vision.AlignmentIcpAlgorithm.POINT_TO_PLANE_ICP)

# Print alignment matrix
print("Alignment matrix:")
print(pcs[0].matrix_to_world())
```
